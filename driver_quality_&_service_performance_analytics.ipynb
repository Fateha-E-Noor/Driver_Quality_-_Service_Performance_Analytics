```python
location_analysis = df.groupby("Pickup Location").agg(
    Bookings=("Booking ID", "count"),
    Driver_Cancellation_Rate=("Is_Driver_Cancelled", "mean"),
    Customer_Cancellation_Rate=("Is_Customer_Cancelled", "mean"),
    Incomplete_Rate=("Is_Incomplete", "mean")
)

location_analysis[
    [
        "Driver_Cancellation_Rate",
        "Customer_Cancellation_Rate",
        "Incomplete_Rate"
    ]
] = (
    location_analysis[
        [
            "Driver_Cancellation_Rate",
            "Customer_Cancellation_Rate",
            "Incomplete_Rate"
        ]
    ] * 100
)

location_analysis.round(2).head()
```

```python
location_analysis["Non_Completion_Rate"] = (
    1 - df.groupby("Pickup Location")["Is_Completed"].mean()
) * 100

location_analysis.round(2).head()
```

```python
import getpass
import os
from sqlalchemy import create_engine

username = os.getenv("PGUSER", "postgres")
password = os.getenv("PGPASSWORD") or getpass.getpass("PostgreSQL password: ")
host = "127.0.0.1"
port = os.getenv("PGPORT", "5432")
database = os.getenv("PGDATABASE", "driver_quality")

engine = create_engine(
    f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}"
)

table_name = "uber_data_cleaned"

df.to_sql(
    table_name,
    engine,
    if_exists="replace",
    index=False
)

print(f"Data successfully loaded into table '{table_name}' in database '{database}'.")
```
